# Walkthrough: entanglement distillation as an MDP

Builds a small MDP, solves it exactly with value iteration ($\gamma=1$), and compares the optimal policy against the pumping, nested, and greedy baselines.

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))

from src.mdp import build_mdp
from src.optimal_policy import value_iteration
from src.baselines import build_pumping_policy, build_nested_policy, build_greedy_policy
from src.policy_eval import expected_waiting_time
```

```python
p, m, f0, delta = 0.9, 4, 0.75, 0.04
mdp = build_mdp(p, m, f0, delta)
print(f"Reachable fidelity levels K = {mdp['F']}")
```

```python
V, policy = value_iteration(mdp)
init_id = mdp['state2id'][tuple([0]*m)]
T_opt = -V[init_id]
print(f"T_opt = {T_opt:.4f}")
```

```python
for name, builder in [('pumping', build_pumping_policy), ('nested', build_nested_policy), ('greedy', build_greedy_policy)]:
    policy_b = builder(mdp)
    T = expected_waiting_time(mdp, policy_b)  # value iteration, gamma=1 -- same method used for the optimal policy
    print(f"{name:8s} T = {T:.4f}  (advantage of optimal: {(T-T_opt)/T*100:5.1f}%)")
```
